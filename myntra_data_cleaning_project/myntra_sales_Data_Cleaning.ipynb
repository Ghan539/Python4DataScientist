```python
# remove the ₹ and " " from the initial and final_price column
myntra['initial_price'] = myntra['initial_price'].str.replace('₹','',regex=False)
myntra['final_price'] = myntra['final_price'].str.replace('₹','',regex=False)
```

```python
# remove  " "  and "," from the initial and final_price column

myntra['initial_price'] = myntra['initial_price'].str.replace('"','',regex=False).str.replace(',','',regex=False)
myntra['final_price'] = myntra['final_price'].str.replace('"','',regex=False).str.replace(',','',regex=False)
```

```python
myntra.info()
```

```python
myntra.head(1)
```

```python
# astype change of initial and final price
myntra['initial_price'] = myntra['initial_price'].astype(float)
myntra['final_price'] = myntra['final_price'].astype(float)
myntra.info()
```

```python
# create new discount column
myntra['final_discount'] = myntra['initial_price'] - myntra['final_price']
```

```python
myntra.head(1)
```

```python
# change name of discount column to  discount_per
myntra.rename(columns={'discount': 'discount_per'},inplace=True)
```

```python
# change column position
col=myntra.pop('final_discount')
myntra.insert(7,'discount',col)
```

```python
# change column position

col=myntra.pop('discount_per')
myntra.insert(9,'discount_per',col)
```

```python
col=myntra.pop('final_price')
myntra.insert(7,'final_price',col)
```

```python
myntra.columns
```

```python
# drop columns
myntra.drop(columns=['url','product_details','breadcrumbs'],inplace=True)
```

```python
myntra['discount_per2'] = ((myntra['initial_price'] - myntra['final_price'])/myntra['initial_price'])*100
```

```python
# change column position
col=myntra.pop('discount_per2')
myntra.insert(9,'discount_per2',col)
```

```python
# 25.005815 → 25.0
# round the values

myntra['discount_per2'] = myntra['discount_per2'].round(1)
```

```python
myntra[myntra['discount_per2'] == 0.0].shape
```

```python
# no of discount_per col with zero discount but value assigned is NAN
myntra.loc[(myntra['initial_price'] == myntra['final_price']) & (myntra['discount_per'].isnull()),'discount_per'] = 0.0
```
