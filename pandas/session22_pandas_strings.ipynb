```python
t['lastname']=tit.str.split(',').str.get(0)

```

```python
t[['title','first_name']]=tit.str.split(',').str.get(1).str.strip().str.split(' ',n=1,expand=True)
t.head()
```

```python
t['title'].value_counts()
```

```python
# replace
# miss====ms===mlle



t['title'] = t['title'].str.replace('Ms.','Miss.')

t['title'] = t['title'].str.replace('Mlle.','Miss.')

t['title'].value_counts()


```

```python
t['title'].value_counts()
```

```python
# filtering
# startswith/endswith
t[t['first_name'].str.startswith('B')]['first_name']
```

```python
t[t['first_name'].str.endswith('B')]['first_name']
```

```python
#
t[t['first_name'].str.isdigit()]['first_name']
```

```python
# applying regex
# contains
# search john -> both case
# case= false means case sensitivity is false

t[t['first_name'].str.contains('john',case=False)]['first_name']
```

```python
# find lastnames with start and end char vowel

t[t['lastname'].str.contains('^[aeiouAEIOU].+[aeiouAEIOU]$')]['first_name']
```

```python
# find lastnames with start and end char vowel

t[t['lastname'].str.contains('^[^aeiouAEIOU].+[^aeiouAEIOU]$')]['first_name']
```
