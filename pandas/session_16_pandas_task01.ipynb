###`Q-2:` Write a Pandas program to add, subtract, multiple and divide two Pandas Series.

````python
ps1=pd.Series([11,22,33,44,55,66,77],dtype='int32')
ps2=pd.Series([10,20,30,40,50,60,70],dtype='int32')
print(ps1+ps2)
print(ps1-ps2)
print(ps1*ps2)
print(ps1%ps2)
````
