```python
import numpy as np
import pandas as pd

# Series from lists

country=['india','pakistan','usa','nepal','srilanka']
print(pd.Series(country))

```

```python
# integers
runs=[13,24,56,78,100]
runs_series=pd.Series(runs)
print(runs_series)
```

```python
# .custom index
marks=[67,57,89,100]
subjects=['evs','english','science','hindi']
ss=pd.Series(marks,index=subjects,name='gshyam')
print(ss)
```

```python
# Series from dict
marks={'marks':67,
       'english':57,
       'science':89,
       'hindi':100
       }
marks_series=pd.Series(marks,name='ghanshyam')
print(marks_series)
```
