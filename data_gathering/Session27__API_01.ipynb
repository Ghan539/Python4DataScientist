```python
import requests
apii='48e8ca3ef267a04254b5cf4cde8514ef'
rq=requests.get(f'https://api.themoviedb.org/3/movie/top_rated?api_key={apii}&language=en-US&page=1')
my_data=rq.json()

df6=pd.DataFrame(my_data['results'])[['id','title','release_date','overview','vote_average','vote_count']]
```

```python
my_df=pd.DataFrame()
my_df

for i in range(1,401):
  rq=requests.get(f'https://api.themoviedb.org/3/movie/top_rated?api_key={apii}&language=en-US&page={i}')
  mydata = rq.json()
  temp_df=pd.DataFrame(mydata['results'])[['id','title','release_date','overview','vote_average','vote_count']]
  my_df = pd.concat([my_df,temp_df],ignore_index=True)
```

```python
my_df.to_csv('Gshyam_dataset_api.csv')
```
