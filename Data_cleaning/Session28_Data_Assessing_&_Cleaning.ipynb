```python
# step_01 :
# make a cpoy of dataframes

patients_co=patients.copy()
treatments_co=treatments.copy()
adverse_reactions_co=adverse_reactions.copy()
treatments_cut_co=treatments_cut.copy()
```

```python
# step_02 : missing values

# patients_co[patients_co['address'].isnull()]
patients_co.fillna('no data',inplace=True)
```

```python
patients_co.info()
```

```python
# msiisngvalues in change column

treatments_co['hba1c_change'] = treatments_co['hba1c_start'] -  treatments_co['hba1c_end']


# msiisngvalues in change column in cut table

treatments_cut_co['hba1c_change'] = treatments_cut_co['hba1c_start'] -  treatments_cut_co['hba1c_end']
```

```python
treatments=pd.concat([treatments_co,treatments_cut_co])
```

```python
treatments.head(1)
```

```python
# auralin and novdro using melt()
treatments=treatments.melt(id_vars=['given_name','surname','hba1c_start','hba1c_end','hba1c_change'],var_name='type',value_name='dosage_range')
```

```python
treatments = treatments[treatments['dosage_range'] != '-']
treatments.head()
```

```python
# new columns for dosage columns
treatments['dosage_start'] = treatments['dosage_range'].str.split('-').str.get(0)
treatments['dosage_end'] = treatments['dosage_range'].str.split('-').str.get(1)
```

```python
treatments.head()
```

```python
treatments.drop(columns='dosage_range',inplace=True)
treatments.head(5)
```

```python
# replace u in dose

treatments['dosage_start'] = treatments['dosage_start'].str.replace('u','')
treatments['dosage_end'] = treatments['dosage_end'].str.replace('u','')
```

```python
treatments.head(5)
```

```python
treatments.drop_duplicates(subset=['given_name','surname'],keep='first',inplace=True)
```

```python
treatments.duplicated().sum()
```

```python
treatments['dosage_start'] = treatments['dosage_start'].astype(int)
treatments['dosage_end'] = treatments['dosage_end'].astype(int)
```

```python
treatments=treatments.merge(adverse_reactions_co,how='left',on=['given_name','surname'])
```
