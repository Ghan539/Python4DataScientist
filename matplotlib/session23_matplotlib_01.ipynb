```python
plt.plot(batsman['index'],batsman['V Kohli'])
plt.plot(batsman['index'],batsman['RG Sharma'])

plt.title('rohit vs v kohli career growth')
plt.xlabel('season')
plt.ylabel('runs scored')

```

```python
# color change
plt.plot(batsman['index'],batsman['V Kohli'],color='red')
plt.plot(batsman['index'],batsman['RG Sharma'],color='green')

plt.title('rohit vs v kohli career growth')
plt.xlabel('season')
plt.ylabel('runs scored')
```

```python
# linestyle--dashed,dotted,dashdot

plt.plot(batsman['index'],batsman['V Kohli'],color='red',linestyle='dashed')
plt.plot(batsman['index'],batsman['RG Sharma'],color='green',linestyle='dashed')

plt.title('rohit vs v kohli career growth')
plt.xlabel('season')
plt.ylabel('runs scored')
```

```python
# linewidth=5

plt.plot(batsman['index'],batsman['V Kohli'],color='red',linestyle='dashed',linewidth=7)
plt.plot(batsman['index'],batsman['RG Sharma'],color='green',linestyle='dashed',linewidth=2)

plt.title('rohit vs v kohli career growth')
plt.xlabel('season')
plt.ylabel('runs scored')
```

```python
# ṃarker='D'


plt.plot(batsman['index'],batsman['V Kohli'],color='red',linestyle='dashed',linewidth=2,marker='D')
plt.plot(batsman['index'],batsman['RG Sharma'],color='green',linestyle='dashed',linewidth=2,marker='D')

plt.title('rohit vs v kohli career growth')
plt.xlabel('season')
plt.ylabel('runs scored')
```

```python
# ṃarker='o'


plt.plot(batsman['index'],batsman['V Kohli'],color='red',linestyle='dashed',linewidth=2,marker='o')
plt.plot(batsman['index'],batsman['RG Sharma'],color='green',linestyle='dashed',linewidth=2,marker='o')

plt.title('rohit vs v kohli career growth')
plt.xlabel('season')
plt.ylabel('runs scored')
```

```python
# line labelling
# label='virat'
# plt.legend()

plt.plot(batsman['index'],batsman['V Kohli'],color='red',linestyle='dashed',linewidth=2,marker='D',label='virat')
plt.plot(batsman['index'],batsman['RG Sharma'],color='green',linestyle='dashed',linewidth=2,marker='D',label='r sharma')

plt.title('rohit vs v kohli career growth')
plt.xlabel('season')
plt.ylabel('runs scored')
plt.legend()
```
