```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as  sns
```

colored scatterplots

```python
iris=pd.read_csv('/content/iris.csv')
iris.head()
```

```python
plt.scatter(iris['SepalLengthCm'],iris['PetalLengthCm'])
plt.xlabel('sepal length')
plt.ylabel('petal length')
```

```python
iris.head()
```

```python
iris['Species']=iris['Species'].replace({'Iris-setosa':0 , 'Iris-versicolor':1 , 'Iris-virginica':2})
iris['Species'].sample(10)
```

```python
plt.scatter(iris['SepalLengthCm'],iris['PetalLengthCm'],c=iris['Species'])
plt.xlabel('sepal length')
plt.ylabel('petal length')
```

```python
# colorbar
plt.scatter(iris['SepalLengthCm'],iris['PetalLengthCm'],c=iris['Species'])
plt.xlabel('sepal length')
plt.ylabel('petal length')

plt.colorbar()
```

```python
# cmap and alpha
plt.scatter(iris['SepalLengthCm'],iris['PetalLengthCm'],c=iris['Species'],cmap='jet',alpha=0.7)
plt.xlabel('Sepal Length')
plt.ylabel('Petal Length')
plt.colorbar()
```

plot size


```python
# plt.figure(figsize=(15,7))

plt.figure(figsize=(15,7))
plt.scatter(iris['SepalLengthCm'],iris['PetalLengthCm'],c=iris['Species'],cmap='jet',alpha=0.7)
plt.xlabel('Sepal Length')
plt.ylabel('Petal Length')
plt.colorbar()
```

```python
# annotations
batters=pd.read_csv('/content/batter (1).csv')
batters.head()
```

```python
sp=batters.head(100).sample(25,random_state=5)

plt.scatter(sp['avg'],sp['strike_rate'])
```

```python
# using anotation
x=[1,2,3,4]
y=[5,6,7,8]
plt.scatter(x,y)
plt.text(1,5,'point-A')
plt.text(2,6,'point-A')
plt.text(3,7,'point-A')
plt.text(4,8,'point-A')
```

```python
sp=batters.head(100).sample(25,random_state=5)



plt.figure(figsize=(14,10))
plt.scatter(sp['avg'],sp['strike_rate'])

for i in range(sp.shape[0]):
  plt.text(sp['avg'].values[i],sp['strike_rate'].values[i],sp['batter'].values[i])
```

```python
sp=batters.head(100).sample(25,random_state=5)



plt.figure(figsize=(14,10))
plt.scatter(sp['avg'],sp['strike_rate'],s=sp['runs'])

for i in range(sp.shape[0]):
  plt.text(sp['avg'].values[i],sp['strike_rate'].values[i],sp['batter'].values[i])
```

```python
# horizontal and vertical lines on graphs
# plt.axhline(130,color='red')

plt.figure(figsize=(14,10))
plt.scatter(sp['avg'],sp['strike_rate'],s=sp['runs'])

plt.axhline(130,color='red')
for i in range(sp.shape[0]):
  plt.text(sp['avg'].values[i],sp['strike_rate'].values[i],sp['batter'].values[i])
```

```python
# horizontal and vertical lines on graphs
# plt.axhline(130,color='red')

plt.figure(figsize=(14,10))
plt.scatter(sp['avg'],sp['strike_rate'],s=sp['runs'])

plt.axhline(130,color='red')
plt.axvline(30,color='red')

for i in range(sp.shape[0]):
  plt.text(sp['avg'].values[i],sp['strike_rate'].values[i],sp['batter'].values[i])
```

```python
plt.figure(figsize=(15,6))
plt.scatter(batters['avg'],batters['strike_rate'])
plt.title('Something')
plt.xlabel('Avg')
plt.ylabel('Strike Rate')
```

```python


fig,ax=plt.subplots(figsize=(15,6))
ax.scatter(batters['avg'],batters['strike_rate'])
ax.set_title('something')
ax.set_xlabel('avg')
ax.set_ylabel('strike_rate')
fig.show()
```

```python
# subplots

fig,ax=plt.subplots(nrows=2,ncols=1)
ax[0].scatter(batters['avg'],batters['strike_rate'],color='green')
ax[1].scatter(batters['avg'],batters['runs'],color='red')

ax[0].set_title('avg vs strike rate')
ax[0].set_xlabel('avg')
ax[0].set_ylabel('strike_rate')

ax[1].set_title('avg vs runs')
ax[1].set_xlabel('avg')
ax[1].set_ylabel('runs')
```

```python
fig,ax=plt.subplots(nrows=2,ncols=2,figsize=(10,10))

ax[0,0].scatter(batters['avg'],batters['strike_rate'])
ax[0,1].scatter(batters['avg'],batters['runs'])
ax[1,0].hist(batters['avg'])
ax[1,1].hist(batters['runs'])



# ax[0,0].set_title('avg vs strike rate')
# ax[0,0].set_xlabel('avg')
# ax[0,0].set_ylabel('strike_rate')

# ax[1].set_title('avg vs runs')
# ax[1].set_xlabel('avg')
# ax[1].set_ylabel('runs')
```
