```python
num=int(input("Enter number :"))

def is_even(num):
  """
  this function returns if a given number is odd or even
  """
  if num%2==0:
    return "even"
  else:
    return "odd"

x=is_even(num)
print(x)

print(is_even.__doc__)
print(print.__doc__ )
```

```python
def power(a=1,b=1):
  return a**b

power(2)

print(power(3,2))
print(power(b=2,a=9 ))
```

```python
# *args

def multiply(a,b,c):
  return a*b*c


multiply(2,2,3)
```

```python
# *args

def multiply (*args):
  prod=1
  for i in args:
    prod=prod*i
  return prod

multiply(2,3,4)
```

```python
# *args

def multiply (*others):
  prod=1
  for i in others:
    prod=prod*i
  return prod

multiply(2,3,4)
```

```python
# **kwargs

def display(**kwargs):

  for (key,value) in kwargs.items():
    print(key,'->',value)

```

```python
display(india='delhi',srilanka='colombo',nepal='kathmandu',pak='islamabad')
display(sub="gh")
```
