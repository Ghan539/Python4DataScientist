```python
# numpy
import numpy as np
a=np.array([1,2,3,4,5,6])

print(a)

# 2d array (metric)

b=np.array([[2,4],[24,64]])
print(b)

# 3d array

c=np.array([[[23,34,54,56],[23,45,65,23],[67,79,98,99],[34,56,78,87]]])
print(c)


# dtype
np.array([22,34,45],dtype=complex)


```

```python

d1=np.random.random((3,3))
d1=np.round(d1*100 )
print(d1)

# max
print(np.max(d1))

#min
print(np.min(d1))

# prod
print(np.prod(d1))

# sum
print(np.sum(d1))

# rows max
# 0 col , 1 row


print(np.max(d1,axis=1))
print(np.max(d1,axis=0))
```

```python
ar1=np.arange(1,10).reshape(3,3)
print(ar1)

print(np.max(ar1))
print(np.min(ar1))
print(np.prod(ar1))
print(np.sum(ar1))

print("row",np.max(ar1,axis=1))
print("column",np.max(ar1,axis=0))
print("column",np.max(ar1,axis=0))
print("prod",np.prod(ar1,axis=0))

# trigonometric functions

print(np.sin(ar1))

# dot product

arr1=np.arange(12).reshape(3,4)
arr2=np.arange(12).reshape(4,3)
print(arr1)
print(arr2)
print(np.dot(arr1,arr2))


#   log and exponents

arr3=([[4.72783947e+18, 1.44625706e+12, 6.83767123e+30],
       [5.32048241e+11, 2.45124554e+40, 4.31123155e+15],
       [2.90488497e+13, 6.56599691e+07, 1.09663316e+03]])

r=arr3.reshape((3,3))
print(r)
```

```python
import numpy as np


# array indexing

x1=np.arange(10)
x2=np.arange(12).reshape((3,4))
x3=np.arange(8).reshape((2,2,2))

print(x1)
print(x2)
print(x3)


print(x1[2])
print(x2[1,1])
print(x3[0,1,1])
print(x3[1,1,0])

print(x1[1:5])


# slicing
print(x2)
# 1st row
print(x2[0,: ])

# 1st column
print(x2[:,0])

#3rd column
print(x2[:,2])

# 5,6
# 9,10
print(x2[1:,1:3])

# 0      3
# 8      11

print(x2[0::2,0::3])

#  1    3
#  9    11

print(x2[::2,1::2])

# 4    7

print(x2[1,::3])


# 1 2 3
# 5 6 7

print(x2[:2,1:])

# 1   3
# 5   7

print(x2[0:2,1::2])


```
